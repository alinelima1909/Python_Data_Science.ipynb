### Busca em Profundidade (DFS - Depth-First Search)

```python
print("Iniciando Busca em Profundidade (DFS) a partir do nó 1:")

# nx.dfs_edges retorna um gerador de arestas na árvore DFS
dfs_edges = list(nx.dfs_edges(grafo, source=1))

# Para ver a ordem dos nós visitados, podemos construir uma lista a partir das arestas
visited_nodes_dfs = []
if dfs_edges:
    # O primeiro nó visitado é sempre a 'source'
    visited_nodes_dfs.append(dfs_edges[0][0])
    for u, v in dfs_edges:
        if v not in visited_nodes_dfs:
            visited_nodes_dfs.append(v)
else:
    # Se não houver arestas do source, apenas o source é visitado
    if 1 in grafo.nodes(): # Verifica se o nó 1 existe
        visited_nodes_dfs.append(1)

print(f"Ordem dos nós visitados (DFS): {visited_nodes_dfs}")
print(f"Arestas percorridas na busca DFS: {dfs_edges}")

# Outra forma de visualizar a árvore DFS diretamente
dfs_tree = nx.dfs_tree(grafo, source=1)
print("\nNós na árvore DFS:")
for node in dfs_tree.nodes():
    print(f"Nó: {node}")

print("Arestas na árvore DFS:")
for u, v in dfs_tree.edges():
    print(f"Aresta: ({u}, {v})")
```

### Busca em Largura (BFS) com Controle de Nó Pai e Nós Visitados

```python
from collections import deque

def bfs_with_parents(graph, start_node):
    # Dicionário para armazenar o pai de cada nó
    parents = {start_node: None}
    # Conjunto para rastrear nós visitados (controle de 'imagens' já processadas)
    visited = {start_node}
    # Fila para a BFS
    queue = deque([start_node])

    # Lista para armazenar a ordem de visitação
    visited_order = [start_node]

    print(f"Iniciando BFS a partir do nó {start_node} (com controle de pai e visitados)")

    while queue:
        current_node = queue.popleft()

        # Para cada vizinho do nó atual
        for neighbor in graph.neighbors(current_node):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
                parents[neighbor] = current_node # Registra o nó atual como pai do vizinho
                visited_order.append(neighbor)
                print(f"Visitando nó {neighbor}, pai: {current_node}")
            else:
                print(f"Nó {neighbor} já visitado (controle de 'imagens' ativado), ignorando...")

    return parents, visited_order

# Executar a BFS modificada a partir do nó 1
parents_map, order_visited = bfs_with_parents(grafo, start_node=1)

print("\n--- Resultados da BFS ---")
print(f"Ordem de visitação: {order_visited}")
print("Nós e seus pais:")
for node, parent in parents_map.items():
    if parent is None:
        print(f"Nó {node}: (Raiz)")
    else:
        print(f"Nó {node}: Pai = {parent}")

# Exemplo de como testar o pai de um vértice específico
node_to_test = 4
if node_to_test in parents_map:
    parent_of_test_node = parents_map[node_to_test]
    if parent_of_test_node is None:
        print(f"\nO nó {node_to_test} é o nó inicial da busca e não tem pai direto nesta execução.")
    else:
        print(f"\nO pai do nó {node_to_test} é: {parent_of_test_node}")
else:
    print(f"\nO nó {node_to_test} não foi alcançado pela BFS a partir do nó inicial.")

node_to_test = 7
if node_to_test in parents_map:
    parent_of_test_node = parents_map[node_to_test]
    if parent_of_test_node is None:
        print(f"\nO nó {node_to_test} é o nó inicial da busca e não tem pai direto nesta execução.")
    else:
        print(f"O pai do nó {node_to_test} é: {parent_of_test_node}")
else:
    print(f"O nó {node_to_test} não foi alcançado pela BFS a partir do nó inicial.")
```

### Algoritmo de Dijkstra para Caminho Mais Curto

O algoritmo de Dijkstra é usado para encontrar os caminhos mais curtos entre nós em um grafo. Ele funciona 'construindo' o caminho mais curto a partir do nó de origem para outros nós no grafo, considerando os pesos (neste caso, 'distance') das arestas.

Para o nosso grafo, vamos encontrar o caminho mais curto entre dois nós específicos.

```python
# Definir o nó de origem e o nó de destino para o algoritmo de Dijkstra
start_node_dijkstra = 1
end_node_dijkstra = 4

print(f"Calculando o caminho mais curto do nó {start_node_dijkstra} para o nó {end_node_dijkstra} usando Dijkstra...")

try:
    # Encontrar o caminho mais curto
    shortest_path = nx.shortest_path(grafo, source=start_node_dijkstra, target=end_node_dijkstra, weight='distance')

    # Encontrar o comprimento (custo) do caminho mais curto
    shortest_path_length = nx.shortest_path_length(grafo, source=start_node_dijkstra, target=end_node_dijkstra, weight='distance')

    print(f"\nCaminho mais curto do {start_node_dijkstra} para {end_node_dijkstra}: {shortest_path}")
    print(f"Custo (distância) do caminho mais curto: {shortest_path_length}")

except nx.NetworkXNoPath:
    print(f"\nNão existe caminho do nó {start_node_dijkstra} para o nó {end_node_dijkstra}.")
except Exception as e:
    print(f"Ocorreu um erro: {e}")
```

### Explicação dos Resultados de Dijkstra

*   **Caminho mais curto:** A lista de nós mostra a sequência de nós que você deve seguir do nó de origem ao nó de destino para obter o menor custo total.
*   **Custo (distância) do caminho mais curto:** Este é o valor total das 'distances' (pesos) das arestas ao longo do caminho mais curto encontrado. É o seu 'custo adicional efetivo' mínimo para ir de um ponto ao outro.

Você pode alterar os valores de `start_node_dijkstra` e `end_node_dijkstra` no código acima para calcular o caminho mais curto entre quaisquer outros dois nós do seu grafo.
